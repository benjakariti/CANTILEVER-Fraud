```python
# EDA cell 1: load libraries and the CSV
import pandas as pd
pd.options.display.max_columns = 100

df = pd.read_csv("../data/creditcard.csv")
print("shape:", df.shape)
display(df.head())
print(df.info())
print("Missing values per column:\n", df.isnull().sum())
```

```python
# EDA cell 2: class distribution & simple stats
print(df["Class"].value_counts())
print("Fraud ratio:", df['Class'].mean())  # fraction that are frauds
display(df["Amount"].describe())
```

```python
# EDA cell 3: quick plots
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.countplot(x="Class", data=df)
plt.title("Class balance (0 = normal, 1 = fraud)")
plt.show()

plt.figure(figsize=(8,4))
sns.histplot(df[df['Class']==0]['Amount'], label='normal', bins=50, stat='density', color='C0', alpha=0.6)
sns.histplot(df[df['Class']==1]['Amount'], label='fraud', bins=50, stat='density', color='C1', alpha=0.6)
plt.legend()
plt.title("Transaction Amounts: normal vs fraud")
plt.show()
```
